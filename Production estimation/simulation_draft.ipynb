```julia
# setting up parameters
beta_0 = 1 
beta_l = 0.2 
beta_k = 0.7 
alpha = 0.7 
sigma_eta = 0.2 
sigma_nu = 0.5 
sigma_w = 0.1 
delta = 0.05 
# wage w_t is defined constant at 0.5
```

```julia
function log_production(l,k,omega,eta,beta_0,beta_l,beta_k)
    return beta_0 + beta_l*l + beta_k*k + omega + eta
end 
log_production(1,1,1,1,beta_0,beta_l,beta_k)
```

```julia
function log_labor_choice(k, wage, omega, beta_0, beta_l, beta_k, sigma_eta)
    f(x) = -(log_production(x[1], k, omega, 0, beta_0, beta_l, beta_k) - wage * x[1])
    result = optimize(f, [0.0], LBFGS())
    return Optim.minimizer(result)
end

log_labor_choice(1,1,1, beta_0, beta_l, beta_k, sigma_eta)
```
